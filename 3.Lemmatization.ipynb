### Lemmatization
Lemmatization converts a word into its base (dictionary) form, called the lemma.

Unlike stemming, it doesn't just cut off endings. It uses grammar and a dictionary to find the correct root word.

```python
import nltk

nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

print(lemmatizer.lemmatize("cars"))
print(lemmatizer.lemmatize("studies"))
print(lemmatizer.lemmatize("plates"))
print(lemmatizer.lemmatize("going",pos="v"))
print(lemmatizer.lemmatize("going"))
print(lemmatizer.lemmatize("worst",pos="a"))
print(lemmatizer.lemmatize("finalized",pos="n"))
```
