### Stemming means reducing a word to its root (stem) by removing prefixes or suffixes.

### The stem may or may not be a valid English word ex, playing -> play , played -> play , studies -> studi , etc.

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
print(ps.stem("playing"))
print(ps.stem("studies"))
print(ps.stem("running"))
```

```python
from nltk.stem import LancasterStemmer

ls = LancasterStemmer()

print(ls.stem("running"))
print(ls.stem("studies"))
print(ls.stem("maximum"))
```

```python
from nltk.stem import SnowballStemmer

ss = SnowballStemmer("english")

print(ss.stem("playing"))
print(ss.stem("studies"))
print(ss.stem("goes"))
```

```python
from nltk.stem import RegexpStemmer

stemmer = RegexpStemmer('ing$|ed$|s$', min=5)

print(stemmer.stem("playing"))
print(stemmer.stem("played"))
print(stemmer.stem("snowing"))
```

### For some words, no matter which stemming technique we try, they won't be grammatically correct.
