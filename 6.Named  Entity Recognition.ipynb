Named Entity Recognition (NER) is an NLP technique used to identify and classify important real-world entities in a sentence.

These entities can be:

👤 Person
🌍 Location
🏢 Organization
📅 Date
💰 Money
⏰ Time
📊 Percentage
...and more.

```python
sentence = "Elon Musk founded SpaceX in 2002"
```

```python
import nltk
words=nltk.word_tokenize(sentence)
```

```python
pos_tag=nltk.pos_tag(words)
```

```python
nltk.download("maxent_ne_chunker_tab")
```

```python
nltk.download("words")
```

```python
nltk.ne_chunk(pos_tag).draw()
```
