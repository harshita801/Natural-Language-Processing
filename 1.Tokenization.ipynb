```python
corpus='''NLP (Natural Language Processing) is a branch of Machine Learning and AI that teaches computers to understand and work with human language.
Examples: ChatGPT answering questions! Google Translate translating languages,Gmail spam detection ,Sentiment analysis of reviews ,Voice assistants like Siri and Alexa'''
```

```python
print(corpus)
```

## Tokenization
sentence -> paragraphs


```python
import nltk

nltk.download('punkt_tab')
```

```python
from nltk.tokenize import sent_tokenize
documents=sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
### tokenization, 
## paragraoh ---> words
## sentence -->words

from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
tokenizer=TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
